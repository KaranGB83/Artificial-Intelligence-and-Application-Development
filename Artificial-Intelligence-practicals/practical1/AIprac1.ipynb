```python
# Table-Driven Agent

table = {
    ('A', 'Dirty'): 'Suck',
    ('A', 'Clean'): 'Move Right',
    ('B', 'Dirty'): 'Suck',
    ('B', 'Clean'): 'Move Left'
}

# Take input from user
location = input("Enter Location (A/B): ").strip().upper()
status = input("Enter Status (Dirty/Clean): ").strip().capitalize()

# Find action
percept = (location, status)

if percept in table:
    print("\nPercept:", percept)
    print("Agent Action:", table[percept])
else:
    print("\nInvalid input! Please enter:")
    print("Location: A or B")
    print("Status: Dirty or Clean")
```
