**Given two water jugs of 4 litres and 3 litres capacity, write a Python program
to obtain exactly 2 litres in one jug using Depth-First Search (DFS). [Vary
capacity of jugs]**

```python
# Karan GB 02
```

```python
jug1_capacity = 4
jug2_capacity = 3
target = 2

visited = set()

def dfs(jug1, jug2):
    if (jug1, jug2) in visited:
        return False
    visited.add((jug1, jug2))
    print(jug1, jug2)
    if jug1 == target or jug2 == target:
      print("\nTarget Achieved")
      return True
    # All possible operations
    return (
        dfs(jug1_capacity, jug2) or
        dfs(jug1, jug2_capacity) or
        dfs(0, jug2) or
        dfs(jug1, 0) or
        dfs(max(0, jug1 - (jug2_capacity - jug2)), min(jug2_capacity, jug2+jug1)) or
        dfs(min(jug1_capacity, jug1+jug2), max(0, jug2 - (jug1_capacity - jug1)))
    )

dfs(0, 0)
```
