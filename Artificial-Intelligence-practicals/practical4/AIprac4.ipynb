**Given a weighted graph representing cities and distances between them, write a
Python program to find the least-cost path from Arad to Bucharest using Uniform Cost
Search (UCS). [Provide any other weighted graph for applying UCS].**

```python
# Karan GB 02
```

```python
import heapq  # Used for priority queue (selects lowest cost path first)

# Weighted graph
# Each city connected with other cities and distanmce is the weight
graph = {
    'Mumbai': [('Pune', 150), ('Nashik', 170)],
    'Pune' : [('Mumbai', 150), ('Surat', 350)],
    'Nashik' : [('Mumbai', 170), ('Surat', 250)],
    'Surat' : [('Pune', 350), ('Nashik', 250), ('Ahmedabad', 270)],
    'Ahmedabad' : []
}

def ucs(graph, start, goal):
  # priority queue stores: (total cost, current city, path)
  queue = [(0, start, [start])]

  # stores visited cities
  visited = set()

  while queue:

    # Get cost with the smallest cost
    cost, city, path = heapq.heappop(queue)

    # If destination is reached, return cost and path
    if city == goal:
      return cost, path

    # visit city only once
    if city not in visited:
      visited.add(city)

      # check all connected cities
      for neighbor, distance in graph[city]:

        # Add new path with updated cost
        if neighbor not in visited:
          heapq.heappush(queue, (cost + distance, neighbor, path + [neighbor]))
  # if no path is found
  return None

cost, path = ucs(graph, 'Mumbai', 'Ahmedabad')
print("Least cost path", " -> ".join(path))
print("Total Distance: ", cost, "km")

```
