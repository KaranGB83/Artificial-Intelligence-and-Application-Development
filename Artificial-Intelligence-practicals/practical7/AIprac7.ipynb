**Given a weighted graph representing cities and distances between them, write
a Python program to find the shortest path from Arad to Bucharest using A*
search with a heuristic function (straight-line distance to destination).**

```python
import heapq  # Used for priority queue (selects lowest cost path first)

graph = {
    'Mumbai': [('Pune', 150), ('Nashik', 170)],
    'Pune' : [('Mumbai', 150), ('Surat', 350)],
    'Nashik' : [('Mumbai', 170), ('Surat', 250)],
    'Surat' : [('Pune', 350), ('Nashik', 250), ('Ahmedabad', 270)],
    'Ahmedabad' : []
}

# heuristic values (ESTIMATED DISTANCE TO AHMDABAD)
heuristic = {
    'Mumbai': 520,
    'Pune': 420,
    'Nashik': 330,
    'Surat': 270,
    'Ahmedabad': 0
}

def astar(graph, start, goal):
  queue = [(heuristic[start], 0, start, [start])]
  visited = set()

  while queue:
    f, cost, city, path = heapq.heappop(queue)

    if city == goal:
      return cost, path

    if city not in visited:
      visited.add(city)

      for neighbor, distance in graph[city]:
        if neighbor not in visited:
          new_cost = cost + distance
          new_f = new_cost + heuristic[neighbor]
          heapq.heappush(queue, (new_f, new_cost, neighbor, path + [neighbor]))

  return None

cost, path = astar(graph, 'Mumbai', 'Ahmedabad')
print("Shortest path: ", " -> ".join(path))
print("Total Distance: ", cost, "km")
```
