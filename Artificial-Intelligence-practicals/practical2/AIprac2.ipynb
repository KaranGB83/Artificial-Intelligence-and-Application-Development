**Given an initial configuration of the 8-puzzle and a goal configuration, write a Python program to find the shortest sequence of moves to reach the goal state using Breadth-First Search (BFS).**

```python
# Karan GB 02
```

```python
from collections import deque  # import deque for implementing the queue in BFS

# Initial state of the 8-puzzle
start = (
    1, 2, 3,
    4, 0, 5,
    6, 7, 8
)

# Goal state of the 8-puzzle
goal = (
    1, 2, 3,
    4, 5, 6,
    7, 8, 0
)

# Dictionary showing all possible moves of the blank tile (0)
# Index positions:
# 0 1 2
# 3 4 5
# 6 7 8

moves = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}

# BFS function to find the shortest path
def bfs(start, goal):
    # Queue stores (current_state, path_taken)
    queue = deque([(start, [start])])

    # Set to store visited states
    visited = {start}

    # Continue until queue becomes empty
    while queue:

        # Remove the first state from the queue
        state, path = queue.popleft()

        # Check if goal state is reached
        if state == goal:
            return path  # Return the solution path

        # Find the index of the blank tile (0)
        zero = state.index(0)

        # Generate all possible next states
        for new in moves[zero]:

            # Convert tuple to list so we can modify it
            new_state = list(state)

            # Swap the blank tile (0) with the adjacent tile
            new_state[zero], new_state[new] = (
                new_state[new],
                new_state[zero]
            )

            # Convert list back to tuple
            new_state = tuple(new_state)

            # If the new state has not been visited
            if new_state not in visited:

                # Mark it as visited
                visited.add(new_state)

                # Add the new state and updated path to the queue
                queue.append((new_state, path + [new_state]))

    # Return None if no solution exists
    return None


# Call the BFS function
solution = bfs(start, goal)

# Print the solution steps
print("Solution Steps:")

# Display each state in a 3x3 puzzle format
for step in solution:
    print(step[0:3])
    print(step[3:6])
    print(step[6:9])
    print()
```
