**Using a small dataset (e.g., weather or spam dataset), implement a Naive Bayes
classifier using scikit-learn.**

```python
import pandas as pd
from sklearn.naive_bayes import CategoricalNB

#Load dataset
url = "https://gist.githubusercontent.com/DiogoRibeiro7/c6590d0cf119e87c39e31c21a9c0f3a8/raw/PlayTennis.csv"

data =  pd.read_csv(url)

# Encoding
data["Outlook"] = data["Outlook"].map({
    "Sunny": 0,
    "Overcast": 1,
    "Rain": 2
})

data["Temperature"] = data["Temperature"].map({
    "Hot": 0,
    "Mild": 1,
    "Cool": 2
})

data["Humidity"] = data["Humidity"].map({
    "High": 0,
    "Normal": 1
})

data["Wind"] = data["Wind"].map({
    "Weak": 0,
    "Strong": 1
})

data["Play Tennis"] = data["Play Tennis"].map({
    "Yes": 0,
    "No": 1
})

# Input and Output
X = data.drop("Play Tennis", axis=1)
y = data["Play Tennis"]

model = CategoricalNB()
model.fit(X, y)
# New weather
new_weather = [[2,0,0,1]]

# Prediction
prediction = model.predict(new_weather)

if prediction[0] == 1:
  print("Play Tennis: Yes")
else:
  print("Play Tennis: No")
```
